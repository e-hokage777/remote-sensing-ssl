```python
import os
import json
```

```python
def get_locations_from_json(json_path: str):
    with open(json_path, "r") as f:
        data = json.load(f)
    return data["elements"]
```

```python
files = os.listdir("data/ghana-locations")

for file in files:
    num_locations = len(get_locations_from_json(f"data/ghana-locations/{file}"))
    print(file, num_locations)
```
